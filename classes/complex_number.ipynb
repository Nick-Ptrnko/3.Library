```python
class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __add__(self, other):
        return Complex(
            self.real + other.real,
            self.imag + other.imag)
    def __sub__(self, other):
        return Complex(
            self.real - other.real,
            self.imag - other.imag)
    def __repr__(self):
        return f"({self.real}, {self.imag})"
    def __str__(self):
        return f"({self.real} + {self.imag}i)"
```

```python
number1 = Complex(1, 2)
number2 = Complex(3, 4)
```

```python
number1 + number2
```

```python
str(number1)
```

```python
number1 - number2
```
