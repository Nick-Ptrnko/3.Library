```python
class FirstError(Exception):
    pass


class SecondError(FirstError):
    pass


class ThirdError(SecondError):
    pass


def first():
    raise FirstError("Error in function: first!")


def second():
    try:
        first()
    except FirstError:
        print("Error in function: first!")

    raise SecondError("Error in function: second!")


def third():
    try:
        second()
    except SecondError as error:
        print(error)
    else:
        try:
            raise ThirdError("Error in function: third!")
        except ThirdError as error:
            print(error)
    finally:
        print("The end!")


third()


```
